# Atribuição de Variáveis

## Variáveis

Uma **variável** é um nome que se refere a um valor.
Para criar uma variável, podemos escrever uma **instrução de atribuição** como esta.

````python
n = 17
````

Uma instrução de atribuição tem três partes: o nome da variável à esquerda, o operador de igualdade, `=`, e uma expressão à direita.
Neste exemplo, a expressão é um número inteiro.
No exemplo a seguir, a expressão é um número de ponto flutuante.

````python
pi = 3.141592653589793
````

E, no exemplo a seguir, a expressão é uma string.

````python
mensagem = 'Boa noite!'
````

Ao executar um comando de atribuição, não há saída.
O Python cria a variável e lhe atribui um valor, mas o comando de atribuição não produz nenhum efeito visível.
No entanto, depois de criar uma variável, você pode usá-la como uma expressão.
Assim, podemos exibir o valor de `mensagem` desta forma:

````python
mensagem
````

## Nomes de variáveis

Os nomes de variáveis ​​podem ter qualquer comprimento. Eles podem conter letras e números, mas **não podem começar com um número**.
É permitido usar letras maiúsculas, mas a convenção é utilizar apenas letras minúsculas para nomes de variáveis (*snake_case*).

O único caractere de pontuação permitido em um nome de variável é o sublinhado (`_`). Ele é frequentemente usado em nomes compostos por várias palavras, como `seu_nome` ou `primeiro_numero_inserido`.

Se você atribuir um nome inválido a uma variável, ocorrerá um erro de sintaxe.
O nome `mil!` é inválido porque contém pontuação.

````python
mil! = 1000000
````

`44gatos` é inválido porque começa com um número.

````python
44gatos = 'Hello'
````

````python
class = 'Hello World'
````

Isso ocorre pois `class` é uma **palavra-chave**, ou seja, é uma palavra especial usada para especificar a estrutura de um programa.
Palavras-chave não podem ser usadas como nomes de variáveis.

## Operadores Aritméticos

Um **operador aritmético** é um símbolo que representa um cálculo aritmético. Por exemplo, o sinal de mais, +, realiza uma adição.

````python
30 + 12
````

O sinal de menos, `-`, é o operador que realiza a subtração.


````python
43 - 1
````

O asterisco, `*`, realiza a multiplicação.

````python
6 * 7
````

E a barra, `/`, realiza a divisão:

````python
84 / 2
````

Observe que o resultado da divisão é `42.0` em vez de `42`. Isso ocorre porque existem dois tipos de números no Python:

* **números inteiros** (*integers*), que representam números sem parte fracionária ou decimal, e

* **números de ponto flutuante** (*floating-point numbers*), que representam números inteiros e números com parte decimal.

Se você somar, subtrair ou multiplicar dois números inteiros, o resultado será um número inteiro.
Mas, se você dividir dois números inteiros, o resultado será um número de ponto flutuante.
O Python oferece outro operador, `//`, que realiza a **divisão inteira**.
O resultado da divisão inteira é sempre um número inteiro.

````python
84 // 2
````

A divisão de inteiros também é chamada de "divisão de piso" (*floor division*), pois sempre arredonda para baixo.

````python
85 // 2
````

Por fim, o operador `**` realiza a exponenciação; isto é, ele eleva um
número a uma potência:

````python
7 ** 2
````

Em algumas outras linguagens, o caractere `^` é usado para exponenciação, mas no Python ele é um operador bit a bit chamado XOR. Se você não estiver familiarizado com operadores bit a bit, o resultado pode ser inesperado:

````python
7 ^ 2
````

Bitwise operators: <http://wiki.python.org/moin/BitwiseOperators>.

## Expressões

Um conjunto de operadores e números é chamado de **expressão**.

Uma expressão pode conter qualquer número de operadores e números.

Por exemplo, aqui está uma expressão que contém dois operadores.

````python
6 + 6 ** 2
````

Observe que a exponenciação ocorre antes da adição.
O Python segue a ordem das operações que você provavelmente aprendeu nas aulas de matemática: a exponenciação ocorre antes da multiplicação e da divisão, que ocorrem antes da adição e da subtração.

No exemplo a seguir, a multiplicação ocorre antes da adição.

````python
12 + 5 * 6
````

Se você quiser que a adição ocorra primeiro, pode usar parênteses.

````python
(12 + 5) * 6
````

Toda expressão tem um **valor**.
Por exemplo, a expressão `6 * 7` tem o valor `42`.

## Funções aritméticas

Além dos operadores aritméticos, o Python oferece algumas **funções** que operam com números.
Por exemplo, a função `round` recebe um número de ponto flutuante e o arredonda para o número inteiro mais próximo.

````python
round(42.4)
````

````python
round(42.6)
````

A função `abs` calcula o valor absoluto de um número. Para um número positivo, o valor absoluto é o próprio número.

````python
abs(42)
````

Para um número negativo, o valor absoluto é positivo.

````python
abs(-42)
````

Quando usamos uma função como esta, dizemos que estamos **chamando** a função.

Uma expressão que chama uma função é uma **chamada de função**.

Ao chamar uma função, os parênteses são obrigatórios.

Se você os omitir, receberá uma mensagem de erro.

````python
abs 42
````

Você pode ignorar a primeira linha desta mensagem; ela não contém nenhuma informação que precisamos entender agora.
A segunda linha é o código que contém o erro, com um acento circunflexo (`^`) abaixo dele para indicar onde o erro foi detectado.

A última linha indica que se trata de um **erro de sintaxe**, o que significa que há algo errado com a estrutura da expressão.

Neste exemplo, o problema é que uma chamada de função requer parênteses.

Vamos ver o que acontece se você omitir os parênteses *e* o valor.

````python
abs
````

O nome de uma função por si só é uma expressão válida que possui um valor.
Quando exibido, o valor indica que `abs` é uma função e inclui algumas informações adicionais que explicarei mais tarde.

## Strings

Além de números, o Python também pode representar sequências de letras, que são chamadas de **strings** porque as letras são enfileiradas como contas em um colar.
Para escrever uma string, podemos colocar uma sequência de letras dentro de aspas retas.

````python
'Hello'
````

Também é permitido usar aspas duplas.

````python
"world"
````

Aspas duplas facilitam a escrita de uma string que contenha um apóstrofo, que é o mesmo símbolo de uma aspa simples reta.

````python
"it's a small "
````

Strings também podem conter espaços, pontuação e dígitos.

````python
'Well, '
````

O operador `+` funciona com strings; ele une duas strings em uma única string, o que é chamado de **concatenação**.

````python
'Well, ' + "it's a small " + 'world.'
````

O operador `*` também funciona com strings; ele cria múltiplas cópias de uma string e as concatena.

````python
'Hello, ' * 4
````

Os outros operadores aritméticos não funcionam com strings.

O Python oferece uma função chamada `len` que calcula o comprimento de uma string.

````python
len('Hello')
````

Observe que `len` conta as letras entre as aspas, mas não as próprias aspas.

Ao criar uma string, certifique-se de usar aspas retas.

O acento grave, também conhecido como crase, causa um erro de sintaxe.

````python
`Hello`
````

Aspas inteligentes, também conhecidas como aspas curvas, também são proibidas.

````python
‘Hello’
````

## Valores e Tipos de Dados

Até agora, vimos três tipos de valores:

* `2` é um número inteiro,

* `42.0` é um número de ponto flutuante, e

* `'Hello'` é uma string.

Um tipo de valor é chamado de **tipo**.
Todo valor tem um tipo — ou, às vezes, dizemos que ele "pertence" a um tipo.

O Python fornece uma função chamada `type` que informa o tipo de qualquer valor.
O tipo de um número inteiro é `int`.

````python
type(2)
````

O tipo de um número de ponto flutuante é `float`.

````python
type(42.0)
````

E o tipo de uma string é `str`.

````python
type('Hello, World!')
````

Os tipos `int`, `float` e `str` podem ser usados ​​como funções.
Por exemplo, `int` pode receber um número de ponto flutuante e convertê-lo em um inteiro (sempre arredondando para baixo).

````python
int(42.9)
````

E `float` pode converter um número inteiro em um valor de ponto flutuante.

````python
float(42)
````

Agora, aqui está algo que pode ser confuso.
O que acontece se você colocar uma sequência de dígitos entre aspas?

````python
'126'
````

````python
type('126')
````

Se você tentar usá-lo como um número, poderá receber um erro.

````python
'126' / 3
````

Este exemplo gera um `TypeError`, o que significa que os valores na expressão, chamados de **operandos**, têm o tipo errado.
A mensagem de erro indica que o operador `/` não suporta os tipos desses valores, que são `str` e `int`.

Se você tiver uma string que contenha dígitos, poderá usar `int` para convertê-la em um inteiro.

````python
int('126') / 3
````

Se você tiver uma string que contenha dígitos e um ponto decimal, pode usar `float` para convertê-la em um número de ponto flutuante.

````python
float('12.6')
````

Ao escrever um número inteiro grande, você pode ser tentado a usar vírgulas
entre grupos de dígitos, como em `1.000.000`.

Esta é uma expressão válida em Python, mas o resultado não é um número inteiro.

````python
1,000,000
````

Python interpreta `1.000.000` como uma sequência de números inteiros separados por vírgulas.
Aprenderemos mais sobre esse tipo de sequência posteriormente.

Você pode usar sublinhados para facilitar a leitura de números grandes.

````python
1_000_000
````

## Glossário

**variável:**
Um nome que se refere a um valor.

**comando de atribuição:**
Um comando que atribui um valor a uma variável.

**palavra-chave:**
Uma palavra especial usada para especificar a estrutura de um programa.

**operador aritmético:**
Um símbolo, como `+` e `*`, que indica uma operação aritmética, como adição ou multiplicação.

**inteiro:**
Um tipo que representa números sem parte fracionária ou decimal.

**ponto flutuante:**
Um tipo que representa números inteiros e números com partes decimais.

**divisão inteira:**
Um operador, `//`, que divide dois números e arredonda o resultado para baixo, chegando a um número inteiro.

**expressão:**
Uma combinação de variáveis, valores e operadores.

**valor:**
Um número inteiro, um número de ponto flutuante ou uma string — ou um dos outros tipos de valores que veremos mais adiante.

**função:**
Uma sequência nomeada de instruções que realiza alguma operação útil.
Funções podem ou não receber argumentos e podem ou não produzir um resultado.

**chamada de função:**
Uma expressão — ou parte de uma expressão — que executa uma função.
Consiste no nome da função seguido por uma lista de argumentos entre parênteses.

**erro de sintaxe:**
Um erro em um programa que torna impossível sua análise sintática (*parsing*) — e, portanto, impossível de executar.

**string:**
Um tipo que representa sequências de caracteres.

**concatenação:**
A união de duas strings, uma após a outra.

**tipo:**
Uma categoria de valores.
Os tipos que vimos até agora são inteiros (tipo `int`), números de ponto flutuante (tipo `float`) e strings (tipo `str`).

**operando:**
Um dos valores sobre os quais um operador atua.

**bug:**
Um erro em um programa.

**depuração:**
O processo de encontrar e corrigir erros.
