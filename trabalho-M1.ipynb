### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5
Preço do litro (R$): 5.89
Total de pedágios (R$): 32.40
Número de pessoas: 3

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
# Inclua aqui a resposta

while True:
  try:
    distancia_km = float(input("Distancia da viagem (km): "))
    consumo = float(input("Consumo medio (km/l): "))
    litro = float(input("Preco do litro: R$ "))
    pedagio = float(input("Total de pedagio: R$ "))
    pessoas = int(input("Num de pessoas: "))
    break
  except ValueError:
    print("Digite um numero valido!")

def div(x, y):
  """Divide 2 valores e retorna o resultado
  x: float
  y: float
  """
  return x / y

def soma(x, y):
  """Soma 2 valores e retorna o resultado
  x: float
  y: float
  """
  return x + y

def custo_combustivel(distancia, consumo, litro):
  """Calcula o custo do combustivel
  Divide a distancia pelo consumo e multiplica pelo preço do litro
  A funcao retorna o resultado dessa conta.

  distancia: float
  consumo: float
  litro: float
  """
  return div(distancia, consumo) * litro

litros = div(distancia_km, consumo)
combustivel = custo_combustivel(distancia_km, consumo, litro)
custo_total = soma(combustivel, pedagio)
custo_pessoas = div(custo_total, pessoas)


print()

print(f"Litros necessarios: {round(litros, 2)}")
print(f"Custo do combustivel: R$ {round(combustivel, 2)}")
print(f"Custo total da viagem: R$ {round(custo_total,2)}")
print(f"Custo por pessoa: R$ {round(custo_pessoas, 2)}")
print(f"Tipo da distancia: {type(distancia_km)} | Tipo do numero de pessoas: {type(pessoas)}")
````

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
# Inclua aqui a resposta
import math


x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float (input("y2: "))

dx = abs(x1 - x2)
dy = abs(y1 - y2)

distancia = math.sqrt(dx ** 2 + dy ** 2)

print(f"dx = {round(dx,2)}")
print(f"dy = {round(dy, 2)}")
print(f"Distancia entre os pontos: {round(distancia, 2)}")
print(f"Area do circulo de raio 5.0: {round(math.pi * (distancia**2), 2)}")
````

### Questão 6: Desenhando com `jupyturtle` (1,0 ponto)

Não esqueça de baixar o módulo `jupyturtle` antes de resolver a questão.


**(a)** Escreva uma função chamada `poligono(lados, tamanho)` que desenhe um polígono regular com o número de lados e o comprimento de lado informados. A função deve ter docstring.

**(b)** Escreva uma função chamada `flor(petalas, lados, tamanho)` que **reutilize** `poligono` para desenhar
uma flor.

**(c)** Faça três desenhos: um triângulo, um hexágono e uma flor.

Exemplo:

```

flor(8, 4, 30)
```

````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py');
````

````python
# (a) e (b) defina aqui as funções poligono e flor
from jupyturtle import left, right, forward, make_turtle

def poligono(lados, tamanho):
  """ Funcao que faz um poligono de acordo com lado e tamanho pedido.
  Forward avanca de acordo com o tamanho desejado, e pra fazer as quinas a tartaruga precisa girar
  um angulo que eh o resultado da divisao do angulo de 360 pelo qtd de lados desejados.

  lados: int
  tamanho: int

  """
  for i in range(lados):
    forward(tamanho)
    left(360 / lados)

def flor(petalas, lados, tamanho):
  """
  Reutiliza a funcao poligono para fazer uma flor. Quando termina um poligono,
  ele volta para o ponto central e comeca outro poligono, em outra direcao, que eh dada pela divisao do angulo 360 pela qtd de petalas. Ele repete isso
  ate terminar de fazer a qtd de petalas pedidas.

  petalas: int
  lados: int
  tamanho: int

  """
  for i in range(petalas):
    poligono(lados, tamanho)
    left(360 / petalas)
````

````python
# (c) triângulo

make_turtle()
poligono(3, 40)
````

````python
# (c) hexágono
make_turtle()
poligono(6, 40)
````

````python
# (c) flor
make_turtle()
flor(5, 5, 30)
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
# Inclua aqui a resposta

peso = float(input("Peso (kg): "))
altura = float(input("Altura (m): "))

def classificar_imc(peso, altura):
  """ Funcao que classifica o imc de acordo com o
  peso e a altura inputados. Valida o valor do peso e da altura (precisam ser > 0),
  calcula o imc e printa a classificacao. A funcao retorna o valor do imc com round.

  peso: float
  altura: float
  imc: float

  """
  if peso <= 0 or altura <= 0:
    print("Dados invalidos: peso e altura devem ser positivos.")
    return None

  imc = peso / altura ** 2

  if imc < 18.5:
    classificacao = "Abaixo do peso"
  elif imc <= 24.9:
    classificacao = "Peso normal"
  elif imc <= 29.9:
    classificacao = "Sobrepeso"
  else:
    classificacao = "Obesidade"

  print(f"IMC: {round(imc, 2)} -> {classificacao}")

classificar_imc(peso, altura)
````

````python
# Versão 1 - somente as funções, testadas com valores fixos

def media_notas(n1, n2, n3):
  """Calcula a media, pega as 3 notas e divide por 3.
  n1: float
  n2: float
  n3: float
  """
  return (n1 + n2 + n3) / 3

media_notas(7, 9, 5)

def situacao(media):
  """Funcao que avalia a situacao do aluno. Se a media do aluno for menor que 4,
  o aluno eh reprovado automaticamente.
  media: float
  """
  if media >= 7:
    return "Aprovado"
  elif media >= 4:
    return "Recuperacao"
  else:
    return "Reprovado"

situacao(4)

def exibir_boletim(nome, media, situacao):
  """
  Exibe o boletim final do aluno.
  nome: string
  media: float
  situacao: string
  """
  print(f"{nome}: {round(media, 1)} | {situacao}")
````
